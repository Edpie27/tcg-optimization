### Series

If Dataframes are tables, then a series is a list of one thing. They are essentially one column of a dataframe. We can create them using the Series function.

```python
Series = pd.Series(['Chocolate', 'Vanilla', 'Strawberry', 'Mint Chocolate Chip'])
Series
```

Like Dataframes, we can assign indexes. We can also name the series.

```python
Series = pd.Series(['Chocolate', 'Vanilla', 'Strawberry', 'Mint Chocolate Chip'], index = ['Best', 'Good', 'Okay','Bad'], name = 'Ice Cream Preferences')
Series
```

### Pandas Indexing

That was indexing using normal python syntax, but pandas also has its own functions for indexing.

```python
df.iloc[0]
```

This retrieves the information from the first row and puts it in a nice series for us.

We can also retrieve a specific value using iloc.

```python
df.iloc[0,3]
```

This retrieves the value from row 0 in column 3 of table df, not the series df.iloc[0]. Do not base your iloc function off this series df.iloc[0] output as then this function above will likely give you an undesired input. Let see what happens if you do look at the series and base it off that.

We want row 4 and column 2 of the series.

```python
df.iloc[4,2]
```

As you can see it gives us some random value. This is because it is based off the df dataframe not this series.

We can retrieve an entire column using iloc.

```python
df.iloc[:,3]
```

The colon tells pandas to read every row and return column 3 only.

We can also use iloc to select a range of values.

```python
df.iloc[:3, 3]
```

The colon infront of the row 3 statement tells pandas to return rows 1-3.

```python
df.iloc[1:3,3]
```

This tells pandas to return just the second and third rows.

We can also use negative numbers in indexing.

```python
df.iloc[-1,3]
```

This returns the last value in the column.

```python
df.iloc[-5:-1,3]
```

This returns the 2nd last value to the 5th last value.

#### loc function vs iloc

loc is used with row or column names while iloc using numerical indexing to determine what row/column to return.

```python
df.loc[:,'name']
```

This returns every row of the 'name' column.

```python
df.iloc[:,3]
```

This iloc returns the exact same output but associates the 'name' column with the third column.

```python
df.loc[0,'name']
```

This returns the first value of the column.

```python
df.iloc[0,3]
```

As does this.

Loc is more intuitive to use in certain cases as the column names are usually representative.

```python
df.loc[:,['kind','name','units','revenue']]
```

This is much easier to remember than the exact positions of the columns.

```python
df.iloc[:,[1,3,4,5]]
```

It is however up to preference and both achieve the same goal reasonably.

You can set the index to a specific column as well.

```python
df.set_index("name")
```

### Conditional Selection

```python
df.revenue >= 50
```

We can return a logical series by inputting a conditonal statement.

```python
df.loc[df.revenue >= 50]
```

We can also nest the conditonal statement in the loc statement to select the relevant data.

We can also use the & statement to bring two questions together.

```python
df.loc[(df.revenue >= 50) & (df.units <= 4)]
```

The | (or) statement also works.

```python
df.loc[(df.revenue >= 50) | (df.units <= 4)]
```

```python
df.loc[df.units.isin([1,2])]
```

We can use the isin function to tell pandas to only retrieve rows with those values.

```python
df.loc[df.kind.notnull()]
```

The notnull/isnull function shows functions that highlights values that are not empty or empty.
