## Pandas Basics

Import the Pandas Library with an alias

```python
import pandas as pd
```

### Dataframes

We can make a dataframe using the DataFrame function.

```python
DataFrame = pd.DataFrame({'Zenless Zone Zero': ['Vivian'], 'Genshin Impact' : ['Raiden Shogun']})
DataFrame
```

Right now, the index is not very descriptive. We can change that by adding a comma and setting index = 'Female'.

```python
DataFrame = pd.DataFrame({'Zenless Zone Zero': ['Vivian'], 'Genshin Impact' : ['Raiden Shogun']}, index = ['Female'])
DataFrame
```

Let us also add another row including male characters.

```python
DataFrame = pd.DataFrame({'Zenless Zone Zero': ['Vivian','Lighter'], 'Genshin Impact' : ['Raiden Shogun','Kaedehara Kazuha']}, index = ['Female','Male'])
DataFrame
```

### Series

If Dataframes are tables, then a series is a list of one thing. They are essentially one column of a dataframe. We can create them using the Series function.

```python
Series = pd.Series(['Chocolate', 'Vanilla', 'Strawberry', 'Mint Chocolate Chip'])
Series
```

Like Dataframes, we can assign indexes. We can also name the series.

```python
Series = pd.Series(['Chocolate', 'Vanilla', 'Strawberry', 'Mint Chocolate Chip'], index = ['Best', 'Good', 'Okay','Bad'], name = 'Ice Cream Preferences')
Series
```

## Reading CSVs

CSV's are a common way to store DataFrames. We can import the sales.csv file by using read_csv.

```python
df = pd.read_csv('sales.csv')
df
```

As you can see it imported a table of all our current sales data including the day of the transaction, the kind of item, the internal key the game stores the item as, the name of the item, the amount of units sold, item revenue, and the highest price it was sold for that day.

We can clean up the default index of 1 through 14 in the first column and change it to the first column in our table.

```python
df_clean = pd.read_csv('sales.csv', index_col = [0])
df_clean
```

In this case, it would probably be best to keep the default index. So we will being using 'df' as the main dataframe.

### Analyze the Table

We can use the .shape function to show how many rows and columns our table has. The output is in the format: (rows, columns).

```python
df.shape
```

We can return the first five rows of the csv using head. We can also input a number into the parenthesis to list that amount of rows.

```python
df.head()
```

```python
df.head(3)
```

We can also do the same with the tail function. It is almost identical to the head function except it returns the last five rows by default. Inputting a number into the parenthesis inputs the last n number of rows instead of the first n number of rows.

```python
df.tail()
```

```python
df.tail(3)
```
