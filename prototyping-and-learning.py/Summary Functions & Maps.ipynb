### Maps

Maps are used to create new representations from existing data.

```python
df_revenue_mean = df.revenue.mean()
dd = df.revenue.map(lambda p: p - df_revenue_mean)
dd
```

We can use the map function to remean the column to zero.

We can also use the apply function to transform an entire dataframe.

```python
def remean_revenue(row):
    row.revenue = row.revenue - df_revenue_mean
    return row

df.apply(remean_revenue, axis = 'columns')
```

This does the exact same thing as the map function. It also creates a new dataframe rather than modifying the original one.

```python
df.head(1)
```

As you can see the revenue column is unmodified.

Pandas also has many common mapping features as built ins.

```python
df_revenue_mean = df.revenue.mean()
df.revenue - df_revenue_mean
```

Pandas understands that we are trying to subtract every value in the series by the mean.
